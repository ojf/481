### Read Noise

Since read noise depends on the CCD and its electronics, it is typically measured from the standard deviation of a bias.

In an aperture, the contribution from read noise is,

$ \sigma_{\text{\sigma_{aperture}}} = \sqrt{n_{pix} \times \sigma_{RN}^2} $

Where $n_{pix}$ is the number of pixels in your aperture, and $\sigma_{RN}$ is the read noise (per pixel) that you measured for your CCD.
